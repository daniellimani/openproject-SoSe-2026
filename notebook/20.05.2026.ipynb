## Which part of my dataset will I focus on today? 

### Generating a new spreadsheet of authors that we are missing their data in one or two fields so I could enhance this data in the next class. 

```python
import pandas as pd

# Load dataset
df = pd.read_csv("artists.csv")

# Select rows where important fields are missing
missing_data_artists = df[
    df["Artist ID"].isnull() |
    df["Name"].isnull() |
    df["Nationality"].isnull() |
    df["Gender"].isnull() |
    df["Birth Year"].isnull()
]

# Display results
print("Artists with missing important information:")
print(missing_data_artists)

# Save to new CSV file
missing_data_artists.to_csv("artists_missing_critical_data.csv", index=False)

print("\nNew spreadsheet created:")
print("artists_missing_critical_data.csv")
```

### I generataed a new spreadsheet containing only artists where one or more of these fields are missing:

#### Artist ID
#### Name
#### Nationality
#### Gender
#### Birth Year

### BUT:

#### missing Death Year is acceptable and should NOT be included as an issue.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("artists.csv")

# Keep only artists with complete important information
complete_artists = df.dropna(
    subset=["Artist ID", "Name", "Nationality", "Gender", "Birth Year"]
)

# Display results
print("Artists with complete important information:")
print(complete_artists)

# Save to new CSV file
complete_artists.to_csv("artists_complete_data.csv", index=False)

print("\nNew spreadsheet created:")
print("artists_complete_data.csv")
```
