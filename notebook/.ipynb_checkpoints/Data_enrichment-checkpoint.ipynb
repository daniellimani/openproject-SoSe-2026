## 4. Data Enrichment Using Wikidata

The purpose of this step is to enrich artist metadata using Wikidata.

For every artist with missing birth year or death year, the notebook searches Wikidata.

If multiple Wikidata entries are returned, the following attributes are compared:

- Name
- Nationality
- Gender
- Birth year
- Death year
- Occupation (artist, painter, sculptor, photographer, etc.)

If one candidate clearly matches the available metadata, the record is marked as **matched**.

If multiple candidates remain possible, the record is marked as **uncertain**.

If no reliable candidate is found, the record is marked as **not_found**.

Only missing values are updated. Existing values are never overwritten.

```python
import pandas as pd
import requests
import time
```

```python
df = pd.read_csv("../data/artists_prepared_for_wikidata.csv")

df.head()
```

```python
import requests

def search_wikidata(name):

    url = "https://www.wikidata.org/w/api.php"

    headers = {
        "User-Agent": "MoMAResearchProject/1.0 (student project)"
    }

    params = {
        "action": "wbsearchentities",
        "search": name,
        "language": "en",
        "format": "json",
        "limit": 5
    }

    response = requests.get(
        url,
        params=params,
        headers=headers,
        timeout=20
    )

    print("Status:", response.status_code)

    if response.status_code != 200:
        print(response.text)
        return None

    return response.json()
```

```python
result = search_wikidata("Pablo Picasso")

result
```

```python
def get_first_candidate(result):

    if len(result["search"]) == 0:
        return None

    return result["search"][0]
```

## Data Enrichment Using Wikidata SPARQL

This step enriches the dataset by querying the Wikidata SPARQL endpoint.

The enrichment process retrieves:

- Wikidata ID
- Birth Year
- Death Year
- Nationality
- Gender
- Occupation

The returned metadata is compared against the existing dataset to decide whether the artist is:

- matched
- uncertain
- not_found

Only missing Birth Year and Death Year values are updated.

```python
import pandas as pd
import requests
import time
```

```python
df = pd.read_csv("../data/artists_prepared_for_wikidata.csv")

df.head()
```

```python
if "wikidata_id" not in df.columns:
    df["wikidata_id"] = None

if "match_status" not in df.columns:
    df["match_status"] = None

if "source" not in df.columns:
    df["source"] = None

df["wikidata_id"] = df["wikidata_id"].astype("object")
df["match_status"] = df["match_status"].astype("object")
df["source"] = df["source"].astype("object")

print("Columns prepared successfully.")
```

```python
def query_wikidata(name):

    query = f"""
    SELECT ?person ?personLabel ?birth ?death ?countryLabel ?genderLabel ?occupationLabel
    WHERE {{
      ?person rdfs:label "{name}"@en.
      OPTIONAL {{ ?person wdt:P569 ?birth. }}
      OPTIONAL {{ ?person wdt:P570 ?death. }}
      OPTIONAL {{ ?person wdt:P27 ?country. }}
      OPTIONAL {{ ?person wdt:P21 ?gender. }}
      OPTIONAL {{ ?person wdt:P106 ?occupation. }}

      SERVICE wikibase:label {{ bd:serviceParam wikibase:language "en". }}
    }}
    LIMIT 5
    """

    url = "https://query.wikidata.org/sparql"

    headers = {
        "Accept": "application/sparql-results+json",
        "User-Agent": "MoMAResearchProject/1.0"
    }

    response = requests.get(
        url,
        params={"query": query},
        headers=headers,
        timeout=20
    )

    return response.json()
```

```python
def calculate_score(row, candidate):

    score = 0

    # Nationality match
    if "countryLabel" in candidate:
        if pd.notna(row["Nationality"]):

            if row["Nationality"].lower() in candidate["countryLabel"]["value"].lower():
                score += 2

    # Gender match
    if "genderLabel" in candidate:
        if pd.notna(row["Gender"]):

            if row["Gender"].lower() in candidate["genderLabel"]["value"].lower():
                score += 1

    # Birth year match
    if "birth" in candidate:
        if pd.notna(row["Birth Year"]):

            wikidata_birth = int(candidate["birth"]["value"][:4])

            if wikidata_birth == int(row["Birth Year"]):
                score += 3

    # Death year match
    if "death" in candidate:
        if pd.notna(row["Death Year"]):

            wikidata_death = int(candidate["death"]["value"][:4])

            if wikidata_death == int(row["Death Year"]):
                score += 2

    # Occupation match
    if "occupationLabel" in candidate:

        occupation = candidate["occupationLabel"]["value"].lower()

        if any(word in occupation for word in [
            "artist",
            "painter",
            "sculptor",
            "photographer",
            "designer"
        ]):
            score += 2

    return score
```

```python
for index, row in df.iterrows():

    name = row["Name"]

    print(f"Processing row {index}: {name}")

    try:

        result = query_wikidata(name)

        candidates = result["results"]["bindings"]

        if len(candidates) == 0:

            df.at[index, "match_status"] = "not_found"

            continue

        best_score = 0
        best_candidate = None

        for candidate in candidates:

            score = calculate_score(row, candidate)

            if score > best_score:

                best_score = score
                best_candidate = candidate

        # If no good match
        if best_score < 2:

            df.at[index, "match_status"] = "uncertain"

            continue

        # Extract Wikidata ID
        wikidata_url = best_candidate["person"]["value"]

        wikidata_id = wikidata_url.split("/")[-1]

        # Fill missing Birth Year
        if pd.isna(row["Birth Year"]):

            if "birth" in best_candidate:

                df.at[index, "Birth Year"] = int(
                    best_candidate["birth"]["value"][:4]
                )

        # Fill missing Death Year
        if pd.isna(row["Death Year"]):

            if "death" in best_candidate:

                df.at[index, "Death Year"] = int(
                    best_candidate["death"]["value"][:4]
                )

        # Add metadata
        df.at[index, "wikidata_id"] = wikidata_id

        df.at[index, "match_status"] = "matched"

        df.at[index, "source"] = "Wikidata"

        time.sleep(0.05)

    except Exception as e:

        print("Error:", e)

        df.at[index, "match_status"] = "error"
```
