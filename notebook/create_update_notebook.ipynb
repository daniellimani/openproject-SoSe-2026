## Descriptive Statistics of the MoMA Dataset

The following code is used to analyze the Museum of Modern Art (MoMA) Collection dataset.

The code was developed using Python and the pandas library in order to generate descriptive statistics and better understand the structure of the dataset.

The code performs the following tasks:
- loads the dataset,
- displays the first rows of the data,
- examines dataset dimensions,
- checks missing values,
- generates descriptive statistics,
- and exports the statistical summary into the DATA folder.

The outputs are printed directly in the notebook to support interpretation and analysis.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("artists.csv")

# Display first rows
print("First 5 Rows:")
print(df.head())

# Dataset dimensions
print("\nDataset Shape:")
print(df.shape)

# Dataset information
print("\nDataset Information:")
print(df.info())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Descriptive statistics
stats = df.describe(include='all')

print("\nDescriptive Statistics:")
print(stats)

# Save statistics
stats.to_csv("artists_statistics.csv")

print("\nStatistics saved as artists_statistics.csv")
```

```python
missing = df.isnull().sum()

print(missing)
```

```python
statistics = pd.DataFrame({

    "Column": df.columns,

    "Missing Values": df.isnull().sum().values,

    "Unique Values": df.nunique().values

})

statistics
```

```python
statistics.to_csv("../data/artists_statistics.csv", index=False)

print("artists_statistics.csv created successfully!")
```

```python
complete = df.dropna(subset=[
    "Artist ID",
    "Name",
    "Nationality",
    "Gender",
    "Birth Year",
    "Death Year"
])

complete.to_csv("../data/artists_all_fields_complete.csv", index=False)

print("Complete artists:", len(complete))
```

```python
missing_dates = df[
    df["Birth Year"].isna() |
    df["Death Year"].isna()
]

missing_dates.to_csv("../data/artists_missing_dates.csv", index=False)

print("Artists needing enrichment:", len(missing_dates))
```

```python
prepared = missing_dates.copy()

prepared["wikidata_id"] = ""

prepared["match_status"] = ""

prepared["source"] = ""

prepared.to_csv("../data/artists_prepared_for_wikidata.csv", index=False)

print("artists_prepared_for_wikidata.csv created!")
```
