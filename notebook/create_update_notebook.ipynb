## Descriptive Statistics of the MoMA Dataset

The following code is used to analyze the Museum of Modern Art (MoMA) Collection dataset.

The code was developed using Python and the pandas library in order to generate descriptive statistics and better understand the structure of the dataset.

The code performs the following tasks:
- loads the dataset,
- displays the first rows of the data,
- examines dataset dimensions,
- checks missing values,
- generates descriptive statistics,
- and exports the statistical summary into the DATA folder.

The outputs are printed directly in the notebook to support interpretation and analysis.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("artists.csv")

# Display first rows
print("First 5 Rows:")
print(df.head())

# Dataset dimensions
print("\nDataset Shape:")
print(df.shape)

# Dataset information
print("\nDataset Information:")
print(df.info())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Descriptive statistics
stats = df.describe(include='all')

print("\nDescriptive Statistics:")
print(stats)

# Save statistics
stats.to_csv("artists_statistics.csv")

print("\nStatistics saved as artists_statistics.csv")
```
